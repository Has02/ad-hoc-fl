```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
log_path = "logs"
```

```python
experiment_old_num = 1
experiment_new_num = 3
log_file_old = log_path + "/log_exp" + str(experiment_old_num) + "_run1.csv"
log_file_new = log_path + "/log_exp" + str(experiment_new_num) + "_run1.csv"
```

```python
experiment_old_num_non = 2
experiment_new_num_non = 4
log_file_old_non = log_path + "/log_exp" + str(experiment_old_num_non) + "_run1.csv"
log_file_new_non = log_path + "/log_exp" + str(experiment_new_num_non) + "_run1.csv"
```

```python
acc_data_old_non = pd.read_csv(log_file_old_non, header=0)
acc_data_new_non = pd.read_csv(log_file_new_non, header=0)
```

```python
acc_data_old = pd.read_csv(log_file_old, header=0)
acc_data_new = pd.read_csv(log_file_new, header=0)
acc_data_old
```

```python
times_file = log_path + "/round_times.csv"
times_data = pd.read_csv(times_file, header=0)

times_data
```

```python
# Just change index to the experiments you want to plot
experiment_old = 0
experiment_new = 2

old_data = times_data.iloc[experiment_old]
new_data = times_data.iloc[experiment_new]

# drop first row
old_data = old_data.drop(old_data.index[0])
new_data = new_data.drop(new_data.index[0])
```

```python
experiment_old_non = 1
experiment_new_non = 3

old_data_non = times_data.iloc[experiment_old_non]
new_data_non = times_data.iloc[experiment_new_non]

# drop first row
old_data_non = old_data_non.drop(old_data_non.index[0])
new_data_non = new_data_non.drop(new_data_non.index[0])
```

```python
old_data
```

```python
old_data.iloc[0]
```

```python
import ast
# in the second column, there is a string that needs to be converted to a list
# this is for each row in the dataframe
for i in range(len(old_data)):
    old_data.iloc[i] = ast.literal_eval(old_data.iloc[i])

for i in range(len(new_data)):
    new_data.iloc[i] = ast.literal_eval(new_data.iloc[i])
```

```python
# in the second column, there is a string that needs to be converted to a list
# this is for each row in the dataframe
for i in range(len(old_data_non)):
    old_data_non.iloc[i] = ast.literal_eval(old_data_non.iloc[i])

for i in range(len(new_data_non)):
    new_data_non.iloc[i] = ast.literal_eval(new_data_non.iloc[i])
```

```python
old_data
```

```python
# There are 100 rows in the dataframe, each row is a device
# The first column is the device index
# The second column is the list of times for each round
# So, the list of times appends a new time for each round to the list

# I want to get the max time for each round over all the devices
# So, I will create a new list of times, where each element is the max time for that round
# Then, I will plot this list

max_times_old = []
max_times_new = []

for i in range(50):
    els_old = old_data.apply(lambda x: x[i])
    els_new = new_data.apply(lambda x: x[i])
    max_times_old.append(max(els_old.values))
    max_times_new.append(max(els_new.values))    
```

```python
max_times_old_non = []
max_times_new_non = []

for i in range(50):
    els_old_non = old_data_non.apply(lambda x: x[i])
    els_new_non = new_data_non.apply(lambda x: x[i])
    max_times_old_non.append(max(els_old_non.values))
    max_times_new_non.append(max(els_new_non.values))
```

```python
max_times_old
```

```python
max_times_new
```

```python
time_old = 2720.274825811386
time_new = 2784.883444070816

time_filler_old = time_old / 50
time_filler_new = time_new / 50

# add the filler times to the max times for each round
for i in range(50):
    max_times_old[i] += time_filler_old
    max_times_new[i] += time_filler_new

print(max_times_old)
print(max_times_new)
```

```python
time_old_non = 2764.443588733673
time_new_non = 2777.300644636154

time_filler_old_non = time_old_non / 50
time_filler_new_non = time_new_non / 50

# add the filler times to the max times for each round
for i in range(50):
    max_times_old_non[i] += time_filler_old_non
    max_times_new_non[i] += time_filler_new_non

print(max_times_old_non)
print(max_times_new_non)
```

```python
# get total time for each experiment
# time_filler is the time that is added to each round to make the total time
# cumsum plus time_filler in each round gives
total_time_old = np.cumsum(max_times_old)
total_time_new = np.cumsum(max_times_new)

print("Total time old: ", total_time_old)
print("Total time new: ", total_time_new)
```

```python
total_time_old_non = np.cumsum(max_times_old_non)
total_time_new_non = np.cumsum(max_times_new_non)

print("Total time old: ", total_time_old_non)
print("Total time new: ", total_time_new_non)
```

```python
# push 0 to the beginning of the list for max times
max_times_old.insert(0, 0)
max_times_new.insert(0, 0)
```

```python
# push 0 to the beginning of the list for max times
max_times_old_non.insert(0, 0)
max_times_new_non.insert(0, 0)
```

```python
plot_times_old = np.cumsum(max_times_old)
plot_times_new = np.cumsum(max_times_new)
```

```python
plot_times_old_non = np.cumsum(max_times_old_non)
plot_times_new_non = np.cumsum(max_times_new_non)
```

```python
# times are the x axis stored in the max_times_old list and max_times_new list
# accuracies are the y axis stored in the acc_data_old and acc_data_new dataframes
# I will plot the accuracies against the times

plt.plot(plot_times_old, acc_data_old['acc'], label='No Heterogeneity')
plt.plot(plot_times_new, acc_data_new['acc'], label='Heterogeneity')
plt.plot(plot_times_old_non, acc_data_old_non['acc'], label='No Heterogeneity - Non IID')
plt.plot(plot_times_new_non, acc_data_new_non['acc'], label='Heterogeneity - Non IID')
plt.xlabel('Time')
plt.ylabel('Accuracy')
plt.title('Accuracy vs Time')
plt.grid()
plt.legend()
plt.show()
```

```python
# get average time per round
average_time_old = np.mean(max_times_old)
average_time_new = np.mean(max_times_new)

average_time_old_non = np.mean(max_times_old_non)
average_time_new_non = np.mean(max_times_new_non)

print("Average time old: ", average_time_old)
print("Average time new: ", average_time_new)

print("Average time old non: ", average_time_old_non)
print("Average time new non: ", average_time_new_non)
```

```python
# plot total loss vs time
plt.plot(total_time_old, acc_data_old['loss'][1:], label='No Heterogeneity')
plt.plot(total_time_new, acc_data_new['loss'][1:], label='Heterogeneity')
plt.plot(total_time_old_non, acc_data_old_non['loss'][1:], label='No Heterogeneity - Non IID')
plt.plot(total_time_new_non, acc_data_new_non['loss'][1:], label='Heterogeneity - Non IID')
plt.xlabel('Time')
plt.ylabel('Total Loss')
plt.title('Total Loss vs Time')
plt.legend()
plt.grid()
plt.show()
```
